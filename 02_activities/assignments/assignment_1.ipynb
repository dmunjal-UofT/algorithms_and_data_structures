````python
import hashlib

def hash_to_range(input_string: str) -> int:
     hash_object = hashlib.sha256(input_string.encode())
     hash_int = int(hash_object.hexdigest(), 16)
     return (hash_int % 3) + 1
input_string = "deepan"
result = hash_to_range(input_string)
print(result)
````

## Question Three: Move All Zeros to End
**Description**  
Given a list of integers, move all zeros to the end while maintaining the relative order of the non-zero elements. 

**Examples**
```python
Input: nums = [0, 1, 0, 3, 12]
Output: [1, 3, 12, 0, 0]
```
```python
Input: nums = [4, 0, 5, 0, 0, 6]
Output: [4, 5, 6, 0, 0, 0]
```


````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    # TODO
    i_pos = 0;                          # Initial starting List index to 0
    for num in nums:                    # Iterating through List
        #print("Current num: ", num)
        if num != 0:                    # Checking for all Non zero elements
            nums[i_pos] = num           # Assign index position to all non zero elements
                                        #print("IF ipos is : ", i_pos, " and num is : ", num)
            i_pos += 1                  # Incrementing Index only for Non zero list elements
    
    while i_pos < len(nums):            # Iterate through length of list from last positon for first zero element
        nums[i_pos] = 0                 # Assigning all zero elements till end of length in list
                                        # print("WHILE ipos is : ", i_pos, " and last non-zero num value was : ", num) 
        i_pos += 1
    pass 
````
