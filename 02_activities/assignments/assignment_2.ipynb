#### Starter Code for Question 3


```python
from typing import List
```

```python
def missing_num(nums: List) -> int:
    missing_nums = []
    for num in range(0,max(nums)):
        if num not in nums:
            missing_nums.append(num)
    return missing_nums
    
```

```python
lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]
missing_num(lst)
```

```python
# GitHUB link:
# https://github.com/kharmandara/algorithms_and_data_structures/pulls
def is_valid_brackets(s: str) -> bool:
    valid_brackets = {'[': ']', '{': '}', '(': ')'}
    stack = []

    for c in s:
        # Case 1: opening bracket
        if c in valid_brackets:
            stack.append(c)
        # Case 2: closing bracket
        else:
            # No opener available
            if not stack:
                return False

            # Get last unmatched opening bracket
            last_open = stack.pop()

            # Check if it matches expected closing bracket
            if valid_brackets[last_open] != c:
                return False

    # If anything is left unmatched, invalid
    return len(stack) == 0
```

```python
#Modified code:

def is_valid_brackets(s: str) -> bool:
    valid_brackets = {'[': ']', '{': '}', '(': ')'}

    # To check for only brackets in a string
    accepted_characters = "[](){}"                      
    stack = []

    #Removing all white spaces
    s= s.replace(" ","")

    for c in s:
        if c not in accepted_characters:
            print("String contains Non Bracket characters")
            return False   
        # Case 1: opening bracket
        if c in valid_brackets:
            stack.append(c)
        # Case 2: closing bracket
        else:
            # No opener available
            if not stack:
                return False

            # Get last unmatched opening bracket
            last_open = stack.pop()

            # Check if it matches expected closing bracket
            if valid_brackets[last_open] != c:
                return False

    # If anything is left unmatched, invalid
    return len(stack) == 0
```
